## Descripción del Método:
El método de punto fijo consiste en transformar la ecuación original $f(x) = 0$ en una ecuación equivalente de la forma $x = g(x)$.

Para garantizar que el método converja hacia la raíz en el intervalo $[1, 2]$, debemos elegir un despeje adecuado. Despejando $x$ del término cuadrático obtenemos:
$$x^2(x+4) = 10 \implies x = \sqrt{\frac{10}{x+4}}$$

Por lo tanto, nuestra función iterativa será:
$$g(x) = \sqrt{\frac{10}{x+4}}$$
A partir de una estimación inicial $p_0$, calculamos las siguientes aproximaciones usando $p_{n+1} = g(p_n)$ hasta cumplir con el criterio de error.

```python
def g(x):
  return np.sqrt(10 / (x + 4))
```

```python
g(0)
```

```python
def pfijo(p0,i):

    lista=[]
    error=[]
#iniciamos bucle for para generar la secuencia de números
    for i in range(i):
        p1=g(p0)
        lista.append(p1)
        e=np.abs(p1-p0)
        error.append(e)
#Acualizamos el valor de p0 para la siguiente iteración del bucle
        p0=p1

    return lista, error
```
